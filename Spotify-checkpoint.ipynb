```python
# First, let's identify unique artists and count plays per artist
artist_plays = spotify['artist_name'].value_counts()
all_artists = len(artist_plays)

# Artists played more than once
replayed_artists = artist_plays[artist_plays > 1]
artist_repeat_count = len(replayed_artists)

# Convert to regular Python integers before formatting
artist_repeat_count_int = int(artist_repeat_count)
all_artists_int = int(all_artists)
percentage = (artist_repeat_count_int / all_artists_int) * 100

# Print the results
print(f'\nArtists replayed at least once: {artist_repeat_count_int:,} out of {all_artists_int:,} ({percentage:.1f}%)')

# For more detailed analysis
# Distribution of plays per artist
print("\nDistribution of plays per artist:")
play_counts = artist_plays.value_counts().sort_index()
for plays, count in play_counts.items():
    print(f"Artists played {plays} time(s): {count:,} ({count/all_artists*100:.1f}%)")

# Time-based analysis to see if user behavior changes over time
spotify['ts'] = pd.to_datetime(spotify['ts'])
spotify['date'] = spotify['ts'].dt.date

# Count unique artists per day
daily_unique_artists = spotify.groupby('date')['artist_name'].nunique()
print(f"\nAverage unique artists per day: {daily_unique_artists.mean():.1f}")

# Calculate the ratio of replays to new artist discoveries over time
artist_first_play = spotify.sort_values('ts').drop_duplicates('artist_name')
spotify['is_first_play'] = spotify.apply(lambda x: True if x.name in artist_first_play.index else False, axis=1)
daily_stats = spotify.groupby('date').agg(
    new_artists=('is_first_play', 'sum'),
    total_plays=('artist_name', 'count')
)
daily_stats['replay_ratio'] = (daily_stats['total_plays'] - daily_stats['new_artists']) / daily_stats['total_plays']
print("\nAverage daily ratio of replays to total plays: {:.1%}".format(daily_stats['replay_ratio'].mean()))
```
