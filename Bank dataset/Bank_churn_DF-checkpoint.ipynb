```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## CUSTORMER CHURN RATE BY COUNTRY

```python
print(' Churn Rate (%) by country: ')
churn_by_country = bank_churn.groupby('Geography')['Exited'].mean()*100
churn_by_country.sort_values(ascending=False)
```

```python
churn_rate_by_country= bank_churn.groupby('Geography')['Exited'].mean()*100
plt.figure(figsize=(8, 5))
sns.set_style('darkgrid')
churn_rate_by_country.plot(kind='bar', color=['blue', 'red', 'green'])
plt.title(' Churn rate (%) by country ')
plt.ylabel(' churn rate (%)')
plt.xlabel('country')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)


#adding values labels untop of bars
for i, v in enumerate(churn_rate_by_country):
    plt.text(i, v + 0.5, f'{v:.1f}%', ha='center')
plt.savefig('Churn_rate_by_country.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
print('The overall churn rate is {:.1f}% '.format(churn_rate))
plt.figure(figsize=(8,5))
sns.set_style('darkgrid')
sns.countplot(data=bank_churn, x='Exited', palette=['orange','blue'], hue='Exited',legend=False)
plt.title('CHURN DISTRIBUTION (0= Stayed, 1= Churned)')
plt.ylabel('Number of Customers')
for i, v in enumerate(bank_churn['Exited'].value_counts()):
    plt.text(i, v + 50, str(v),ha='center')
plt.savefig('CHURN_DISTRIBUTION.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, axes= plt.subplots(2,2,figsize=(10,8))
sns.set_style('darkgrid')
sns.barplot(x='Gender', y='Exited', data=bank_churn, ax=axes[0,0])
axes[0,0].set_title('CHURN RATE BY GENDER')
axes[0,0].set_ylabel('Churn Rate')

sns.barplot(x='NumOfProducts', y='Exited', data=bank_churn, ax=axes[0,1])
axes[0,1].set_title('CHURN RATE BY NUMBER OF PRODUCTS')
axes[0,1].set_ylabel('Churn Rate')

sns.barplot(x='IsActiveMember', y='Exited', data=bank_churn, ax=axes[1,0])
axes[1,0].set_title('CHURN RATE BY ACTIVE MEMBER')
axes[1,0].set_ylabel('Churn Rate')

sns.barplot(x='HasCrCard', y='Exited', data=bank_churn, ax=axes[1,1])
axes[1,1].set_title('CHURN RATE BY CREDIT CARD OWNERSHIP')
axes[1,1].set_ylabel('Churn Rate')

plt.suptitle('Customer Churn Rate by Key Categories', fontsize=16)
plt.tight_layout()
plt.show()
fig.savefig('Churn_rate_by_categories.png', dpi=300, bbox_inches='tight')
```

```python
numerical_col=['Age','Balance', 'CreditScore','Tenure','EstimatedSalary']
for col in numerical_col:
    plt.figure(figsize=(8,5))
    sns.set_style('darkgrid')
    sns.boxplot(x='Exited', y=col, data=bank_churn, palette=['blue','green'], hue='Exited', legend=False)
    plt.title(f'{col} Distribution by churn status', fontsize=14, pad=15)
    plt.xlabel('Churn Status (0=stayed, 1= churned)')
    plt.ylabel(col)
    plt.tight_layout()
    plt.savefig(f'{col}_distribution_by_churn.png', dpi=300, bbox_inches='tight')
    plt.show()
```

## KEY INSIGHTS & BUSINESS RECCOMENDATIONS

### 1. OVERALL BANK OVERVIEW
- Total Customers: **10,000**
- Overall Churn Rate: **20.37%** (2,037 customers left the bank)

### 2. Country Differences (Most Important Finding)
- **Germany** has a significantly higher churn rate (**32.44%**) compared to France(**16.2%**) and spain(**16.7%**).
- Germany customers are the highest risk group.

### 3. Churner Vs Non-Churner profile
- Churners are on average **older** (44 years vs 37.4 years)
- Churners have ** higher balances** on average but lower number of products and lower active membership rate
- IsActiveMember is a very  strong predictor of churn

### 4. Demographics
- Average age: **38.9 years**
- Average Balance: **$76,486**
- Average Credit Score: **651**
- **70.5%** have Credit cards, also **51.5%** are Active Members
- Females churn more than males (**25%** vs **16.5%**)
- Customers with **3+** products have very high churn rates

**BUSINESS RECOMMENDATION**
- Targeted retention campaign for German customers should be created and prioritized.
- Focus on increasing the banks "Active Member" status- This happens to be a key driver of retention, so it will be nice to focus on customer engagement for Active member status.
- Investigate why older customers with high balances are churning from the bank
