```python
climate_events_df['year'] = pd.to_numeric(climate_events_df['year'], errors='coerce')

print(climate_events_df.isnull().sum())
```

```python
events_count = climate_events_df.groupby(['year','event_type']).size().reset_index(name='count')
climate_events_pivot = events_count.pivot(index='year', columns='event_type', values='count').fillna(0)
print(climate_events_pivot.head(10))
```

```python
plt.figure(figsize=(12,6))
total_events_per_year= climate_events_pivot.sum(axis=1)
sns.lineplot(x=total_events_per_year.index, y=total_events_per_year.values, marker='o', linewidth=2.5)
plt.title('Total Extreme climate Events per Year (2003-2006)', fontsize=14)
plt.xlabel('Year')
plt.ylabel('Number of Events')
plt.grid(True)
plt.savefig("Total_extreme_events_trend.png", dpi=300, bbox_inches='tight')
print("saved:Total_extreme_events_trend.png")
plt.show()
```

```python
climate_events_pivot.plot(kind='area', stacked=True, figsize=(12,6), alpha=0.8)
plt.title('Extreme Events by Type Over time')
plt.xlabel('Year')
plt.ylabel('Number of Events')
plt.legend(title='Event Type', bbox_to_anchor=(1.05, 1))
plt.grid(True)
plt.tight_layout()
plt.savefig("Extreme_events_by_country_type_stacked.png", dpi=300, bbox_inches='tight')
print("saved:Extreme_events_by_country_type_stacked.png")
plt.show()
```

```python
carbon= carbon_prices_daily_df.copy()
carbon['date']= pd.to_datetime(carbon['date'])
print(carbon['market'].value_counts())
```

```python
carbon_pivot=carbon.pivot(index='date',columns='market', values='price')
returns=carbon_pivot.pct_change().dropna()
print('Daily Returns Summary:')
print(returns.describe())
```

```python
# correlation matrix
corr_matrix= returns.corr()
plt.figure(figsize=(14,8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Between Carbon Markets (Daily Returns)')
plt.savefig("Carbon_markets_correlation_heatmap.png",dpi=300,bbox_inches='tight')
print("saved:Carbon_markets_correlation_heatmap.png")
plt.show()
```

```python
vol_30d= returns.rolling(30).std() * np.sqrt(252)

vol_30d.plot(figsize=(12,6), linewidth=2)
plt.title('Annualized volatility of carbon markets (30-day rolling)')
plt.ylabel('Volatility')
plt.grid(True)
plt.savefig("Carbon_markets_volatility_30d.png", dpi=300, bbox_inches='tight')
print("saved:Carbon_markets_volatility_30d.png")
plt.show()
```
